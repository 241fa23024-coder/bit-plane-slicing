```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img=cv2.imread('/content/lab5.jpg',cv2.IMREAD_GRAYSCALE)
```

```python
if img is None:
  print("Error :image not found.")
else:
  for i in range(8):
    plt.figsize=(10,5)
    bit_plane=(img>>i)&1
    vis_plane=bit_plane*255
    plt.subplot(3,3,i+2)
    plt.imshow(vis_plane,cmap='gray')
    plt.title(f"Bit Plane {i}")
    plt.axis('off')
  plt.tight_layout()
  plt.show()
```
